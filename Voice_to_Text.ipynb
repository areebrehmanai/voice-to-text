```python
import IPython.display as display
from IPython.display import Audio
import ipywidgets as widgets
import whisper

model = whisper.load_model("base")

uploader = widgets.FileUpload(accept='.mp3,.wav', multiple=False)
display.display(widgets.Label("Upload a .mp3 or .wav file for voice-to-text transcription"))
display.display(uploader)

def on_upload_change(change):
    if uploader.value:
        # NOTE: uploader.value is a tuple/list of dict-like entries in
        # current ipywidgets (8.x), not a dict keyed by filename as in
        # the older 7.x API. Each entry exposes 'name'/'content' directly
        # (no nested 'metadata' key).
        uploaded_file = uploader.value[0]
        filename = uploaded_file['name']
        content = uploaded_file['content']

        with open(filename, 'wb') as f:
            f.write(content)

        print("\U0001F50D Transcribing audio...")
        result = model.transcribe(filename)

        display.display(Audio(filename))
        print("\U0001F4DD Transcription:")
        print(result['text'])

uploader.observe(on_upload_change, names='value')
```
